```python
from pyspark.sql.functions import col, split, to_date
```

```python
# Read the silver table
df_bronze = spark.table("uk_ecommerce.bronze.online_retail")
```

```python
# Handling duplicates and nulls
df_silver = df_bronze.dropDuplicates().dropna()
df_silver = df_silver.dropna()
```

```python
# Create a totalSales column
df_silver = df_silver.withColumn("totalSales", col("Quantity") * col("UnitPrice"))
```

```python
# extract only the date from InvoiceDate column
df_silver = df_silver.withColumn("InvoiceDate", split(col("InvoiceDate"), " ")[0])

# fixing the date format
df_silver = df_silver.withColumn("InvoiceDate", to_date(col("InvoiceDate"), "M/d/yyyy"))
```

```python
# write the silver table
df_silver.write.mode("overwrite").saveAsTable("uk_ecommerce.silver.online_retail")
```
