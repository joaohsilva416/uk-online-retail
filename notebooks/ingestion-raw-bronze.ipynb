```python
# Load data into Spark DataFrame
df_raw = spark.read.format("csv").load("/Volumes/uk_ecommerce/raw/landing/OnlineRetail.csv", header=True, inferSchema=True)
```

```python
# Convert CustomerID column to string
df_raw = df_raw.withColumn("CustomerID", df_raw["CustomerID"].cast("string"))
```

```python
df_raw.write.mode("overwrite").saveAsTable("uk_ecommerce.bronze.online_retail")
```
