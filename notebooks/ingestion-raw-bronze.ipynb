```python
# Load data into Spark DataFrame
df = spark.read.format("csv").load("/Volumes/uk_ecommerce/raw/landing/OnlineRetail.csv", header=True, inferSchema=True)
```

```python
# Convert CustomerID column to string
df = df.withColumn("CustomerID", df["CustomerID"].cast("string"))
```

```python
df.write.mode("overwrite").saveAsTable("uk_ecommerce.bronze.online_retail")
```
